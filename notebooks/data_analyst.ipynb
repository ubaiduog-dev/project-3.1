```python
import sqlite3
import pandas as pd
con=sqlite3.connect(r'C:\Users\ubaid\OneDrive\Desktop\sem-3\project 3.1\data\raw\ipl.db')
```

```python
def q(sql):
    return pd.read_sql_query(sql,con)
```

```python
def run_sql_file(path):
    with open(path, 'r') as f:
        script = f.read()
    con.executescript(script)
    con.commit()
    print('ran', path)
```

```python
q("""
--what is the strike rate of every batter in the death overs

select batter,
round(100.0*sum(batter_runs))/sum(
    case 
        when is_wide_ball=0 and is_no_ball=0 
        then 1
        else 0
    end) as strike_rate

from deliveries
where over_number>=15 and is_super_over=0
group by batter;""")
```
