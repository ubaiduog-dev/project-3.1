```python
q("""
SELECT 
    batter,
    ROUND(
        100.0 * SUM(batter_runs) / SUM(
            CASE
                WHEN is_wide_ball = 0 AND is_no_ball = 0
                THEN 1
                ELSE 0
            END
        ), 2
    ) AS strike_rate
FROM deliveries
WHERE over_number BETWEEN 15 AND 19
  AND is_super_over = 0
GROUP BY batter
ORDER BY strike_rate DESC;
""")
```

```python
df=pd.read_sql("""
-- list of highest six scores in ipl history

  select batter,
         sum (case when batter_runs=6
              then 1
              else 0
              end)as total_sixes
from v_ball
group by batter
order by total_sixes desc
limit 5;
 """, con)


df.plot(kind="bar",
        x="batter",	
        y="total_sixes",
        xlabel="six score batter",
        ylabel="scored total_sixes",
        color=["red", "blue", "green", "orange", "purple",
           "pink", "yellow", "cyan", "brown", "gray"],
        title="highest six scored batters total sixes",

        )
```

```python
df=pd.read_sql(""" 
 --give the list of highest runs scored in ipl history
 --max 5
select batter,sum(batter_runs) as total_runs
from v_ball
group by batter
order by total_runs desc
limit 5;""",con)
df.plot(kind="bar",
        x="batter",
        y="total_runs",
        xlabel="batters",
        ylabel="total runs",
        title="highest runs scored in ipl history")
```

```python
af= pd.read_sql("""
SELECT 
    bowler,
    SUM(bowler_wicket) AS wicket
FROM v_ball
GROUP BY bowler
ORDER BY wicket DESC
limit 10;
""", con)

plt.figure(figsize=(12, 8))

colors = plt.cm.plasma(
    af["wicket"] / af["wicket"].max()
)

bars = plt.barh(
    af["bowler"],
    af["wicket"],
    color=colors
)

plt.xlabel("Total Wickets")
plt.ylabel("Bowler")
plt.title("Total Wickets by Bowler")

plt.gca().invert_yaxis()

# Add wicket values
for bar in bars:
    plt.text(
        bar.get_width() + 0.5,
        bar.get_y() + bar.get_height()/2,
        int(bar.get_width()),
        va="center"
    )

plt.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
ag = pd.read_sql("""
SELECT
    toss_winner,
    COUNT(*) AS toss_won
FROM matches_clean
GROUP BY toss_winner
ORDER BY toss_won DESC
limit 5;
""", con)

plt.figure(figsize=(10, 8))

plt.pie(
    ag["toss_won"],
    labels=ag["toss_winner"],
    autopct="%1.1f%%",
    startangle=90,
    shadow=True
)

plt.title("Toss Wins by Team")

plt.tight_layout()
plt.show()
```
